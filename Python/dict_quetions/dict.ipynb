```python
# 1. Write a Python program to create a dictionary and display all key-value pairs.
# Input {"name": "Amit", "age": 21}
# Expected Output name: Amit
#                age: 21

student = {"name": "Amit", "age": 21}

for key, value in student.items():
    print(key + ":", value)
```

```python
# 2. Write a Python program to add a new key-value pair to a dictionary.
# Input Dictionary = {"name": "Amit"}
# Add: "city": "Kolhapur"
# Expected Output {"name": "Amit", "city": "Kolhapur"}

student = {"name": "Amit"}

student["city"] = "Kolhapur"

print(student)
```

```python
# 3. Write a Python program to update the value of an existing key in a dictionary.
# Input Dictionary = {"name": "Amit", "age": 20}
# Update age = 21
# Expected Output {"name": "Amit", "age": 21}

student = {"name": "Amit", "age": 20}

student["age"] = 21

print(student)
```

```python
# 4. Write a Python program to delete a key from a dictionary.
# Input Dictionary = {"a": 1, "b": 2, "c": 3}
# Delete key = b
# Expected Output {"a": 1, "c": 3}

d = {"a": 1, "b": 2, "c": 3}

del d["b"]

print(d)
```

```python
# 5. Write a Python program to check whether a key exists in a dictionary.
# Input Dictionary = {"name": "Amit", "age": 21}
# Key = age
# Expected Output Key exists = True

student = {"name": "Amit", "age": 21}

key = "age"

result = key in student

print("Key exists =", result)
```

```python
# 6. Write a Python program to print all keys, all values, and all items of a dictionary.
# Input {"a": 10, "b": 20}
# Expected Output Keys = ['a', 'b']
# Values = [10, 20]
# Items = [('a', 10), ('b', 20)]

d = {"a": 10, "b": 20}

keys = list(d.keys())
values = list(d.values())
items = list(d.items())

print("Keys =", keys)
print("Values =", values)
print("Items =", items)
```

```python
# 7. Write a Python program to find the sum of all numeric values in a dictionary.
# Input {"a": 10, "b": 20, "c": 30}
# Expected Output Sum = 60

d = {"a": 10, "b": 20, "c": 30}

total = 0

for value in d.values():
    total += value

print("Sum =", total)
```

```python
# 8. Write a Python program to merge two dictionaries.
# Input D1 = {"a": 1, "b": 2}
# D2 = {"c": 3, "d": 4}
# Expected Output {"a": 1, "b": 2, "c": 3, "d": 4}

D1 = {"a": 1, "b": 2}
D2 = {"c": 3, "d": 4}

D1.update(D2)

print(D1)
```

```python
# 9. Write a Python program to count the frequency of characters in a string using a dictionary.
# Input aabcc
# Expected Output {'a': 2, 'b': 1, 'c': 2}

text = "aabcc"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

```python
# 10. Write a Python program to create a dictionary from two lists: one of keys and one of values.
# Input Keys = ['name', 'age', 'city']
# Values = ['Amit', 21, 'Kolhapur']
# Expected Output {'name': 'Amit', 'age': 21, 'city': 'Kolhapur'}

keys = ['name', 'age', 'city']
values = ['Amit', 21, 'Kolhapur']

d = {}

for i in range(len(keys)):
    d[keys[i]] = values[i]

print(d)
```

```python
# Group words that are anagrams using a dictionary

words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']

anagram_groups = {}

for word in words:
    
    key = ''.join(sorted(word))
    if key not in anagram_groups:
        anagram_groups[key] = []
    
    anagram_groups[key].append(word)

print(anagram_groups)
```

```python
# Create a word-position index

text = "python is easy and python is powerful"

words = text.split()

word_index = {}

for position, word in enumerate(words):
    if word not in word_index:
        word_index[word] = []
    
    word_index[word].append(position)

print(word_index)
```

```python
# Write a Python program to recursively merge two nested dictionaries.
# Input D1 = {'a': 1, 'b': {'x': 10}}
# D2 = {'b': {'y': 20}, 'c': 3}
# Expected Output {'a': 1, 'b': {'x': 10, 'y': 20}, 'c': 3}

def merge_dict(d1, d2):

    for key in d2:

        if key in d1 and isinstance(d1[key], dict) and isinstance(d2[key], dict):
            merge_dict(d1[key], d2[key])

        else:
            d1[key] = d2[key]

    return d1


D1 = {'a': 1, 'b': {'x': 10}}
D2 = {'b': {'y': 20}, 'c': 3}

result = merge_dict(D1, D2)

print(result)
```
