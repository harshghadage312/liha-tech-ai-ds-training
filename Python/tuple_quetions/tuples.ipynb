```python
"""Write a Python program to calculate the element-wise sum of two tuples.
Input Tuple 1 = (1, 2, 3)
Tuple 2 = (4, 5, 6)
Expected Output (5, 7, 9)"""
t1 = (1, 2, 3)
t2 = (4, 5, 6)

result = ()

for i in range(len(t1)):
    result += (t1[i] + t2[i],)

print(result)
```

```python
t = (1, 2, 2, 3, 4, 4, 4)

repeated = ()

for i in t:
    if t.count(i) > 1 and i not in repeated:
        repeated += (i,)

print("Repeated values =", repeated)
```
