```python
# 12. Write a Python program to print the Fibonacci series up to n terms using range().
# Input n = 7
# Expected Output 0 1 1 2 3 5 8

n = 7

a = 0
b = 1

for i in range(n):
    print(a, end=" ")

    c = a + b
    a = b
    b = c
```

```python
# 15. Write a Python program to print all leap years between two given years using range().
# Input Start year = 2000
# End year = 2020
# Expected Output 2000 2004 2008 2012 2016 2020

start = 2000
end = 2020

for year in range(start, end + 1):
    if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
        print(year, end=" ")
```

```python
# 16. Write a Python program to display a star pattern using nested loops and range().
# Input Rows = 4
# Expected Output
# *
# * *
# * * *
# * * * *

rows = 4

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Write a Python program to print all prime numbers between two given numbers using range().
# Input Start = 10
# End = 30
# Expected Output 11 13 17 19 23 29

start = 10
end = 30

for num in range(start, end + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
# Write a Python program to print Pascal's triangle for n rows using nested range() loops.
# Input Rows = 5
# Expected Output
# 1
# 1 1
# 1 2 1
# 1 3 3 1
# 1 4 6 4 1

rows = 5

for i in range(rows):
    value = 1

    for j in range(i + 1):
        print(value, end=" ")
        value = value * (i - j) // (j + 1)

    print()
```
