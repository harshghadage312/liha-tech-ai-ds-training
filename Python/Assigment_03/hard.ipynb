```python
"""Write a Python program to flatten a nested list of any depth into a single list.
Input [1, [2, [3, 4], 5], 6]
Expected Output [1, 2, 3, 4, 5, 6]"""


def flatten(lst):

    result = []

    for item in lst:

        if isinstance(item, list):
            result.extend(flatten(item))

        else:
            result.append(item)

    return result


lst = [1, [2, [3, 4], 5], 6]

print(flatten(lst))
        
      
      
     
         
       
     
    
    

```
