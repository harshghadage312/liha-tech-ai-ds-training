```python
"""Write a Python program to flatten a nested list of any depth into a single list.
Input [1, [2, [3, 4], 5], 6]
Expected Output [1, 2, 3, 4, 5, 6]"""


def flatten(lst):

    result = []

    for item in lst:

        if isinstance(item, list):
            result.extend(flatten(item))

        else:
            result.append(item)

    return result


lst = [1, [2, [3, 4], 5], 6]

print(flatten(lst))
        
      
      
     
         
       
     
    
    

```

```python
# Write a Python program to find the contiguous subarray having the maximum sum.
# Input [-2, 1, -3, 4, -1, 2, 1, -5, 4]
# Expected Output Maximum sum = 6
# Subarray = [4, -1, 2, 1]

arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = arr[0]
subarray = []

for i in range(len(arr)):
    current_sum = 0

    for j in range(i, len(arr)):
        current_sum += arr[j]

        if current_sum > max_sum:
            max_sum = current_sum
            subarray = arr[i:j + 1]

print("Maximum sum =", max_sum)
print("Subarray =", subarray)
```

```python
# Write a Python program to build a list where each element is the product of all other elements, without using division.
# Input [1, 2, 3, 4]
# Expected Output [24, 12, 8, 6]

numbers = [1, 2, 3, 4]

result = []

for i in range(len(numbers)):
    product = 1

    for j in range(len(numbers)):
        if i != j:
            product *= numbers[j]

    result.append(product)

print(result)
```
