```python
# 1. Write a Python program to create a set and display its elements.
# Input {10, 20, 30, 40}
# Expected Output {10, 20, 30, 40} (order may vary)

s = {10, 20, 30, 40}

print(s)
```

```python
# 2. Write a Python program to add and remove elements from a set.
# Input Set = {10, 20, 30}
# Add = 40
# Remove = 20
# Expected Output {10, 30, 40} (order may vary)

s = {10, 20, 30}

s.add(40)
s.remove(20)

print(s)
```

```python
# 3. Write a Python program to find the union of two sets.
# Input A = {1, 2, 3}
# B = {3, 4, 5}
# Expected Output {1, 2, 3, 4, 5}

A = {1, 2, 3}
B = {3, 4, 5}

result = A.union(B)

print(result)
```

```python
# 4. Write a Python program to find the intersection of two sets.
# Input A = {1, 2, 3, 4}
# B = {3, 4, 5, 6}
# Expected Output {3, 4}

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.intersection(B)

print(result)
```

```python
# 5. Write a Python program to find the difference between two sets.
# Input A = {1, 2, 3, 4}
# B = {3, 4, 5}
# Expected Output A - B = {1, 2}

A = {1, 2, 3, 4}
B = {3, 4, 5}

result = A - B

print("A - B =", result)
```

```python
# 6. Write a Python program to find the symmetric difference between two sets.
# Input A = {1, 2, 3}
# B = {3, 4, 5}
# Expected Output {1, 2, 4, 5}

A = {1, 2, 3}
B = {3, 4, 5}

result = A.symmetric_difference(B)

print(result)
```

```python
# 7. Write a Python program to check whether one set is a subset of another set.
# Input A = {1, 2}
# B = {1, 2, 3, 4}
# Expected Output A is a subset of B = True

A = {1, 2}
B = {1, 2, 3, 4}

result = A.issubset(B)

print("A is a subset of B =", result)
```

```python
# 8. Write a Python program to check whether two sets are disjoint.
# Input A = {1, 2, 3}
# B = {4, 5, 6}
# Expected Output The sets are disjoint = True

A = {1, 2, 3}
B = {4, 5, 6}

result = A.isdisjoint(B)

print("The sets are disjoint =", result)
```

```python
# 9. Write a Python program to remove duplicate elements from a list using a set.
# Input [1, 2, 2, 3, 1, 4]
# Expected Output {1, 2, 3, 4}

numbers = [1, 2, 2, 3, 1, 4]

result = set(numbers)

print(result)
```

```python
# 10. Write a Python program to find the maximum and minimum values in a set.
# Input {8, 3, 15, 1, 9}
# Expected Output Maximum = 15, Minimum = 1

s = {8, 3, 15, 1, 9}

maximum = max(s)
minimum = min(s)

print("Maximum =", maximum)
print("Minimum =", minimum)
```

```python
# 11. Write a Python program to convert a string into a set of unique characters.
# Input banana
# Expected Output {'b', 'a', 'n'} (order may vary)

text = "banana"

result = set(text)

print(result)
```

```python
# 12. Write a Python program to count the number of elements in a set.
# Input {10, 20, 30, 40, 50}
# Expected Output Number of elements = 5

s = {10, 20, 30, 40, 50}

count = len(s)

print("Number of elements =", count)
```

```python
# 13. Write a Python program to clear all elements from a set.
# Input {1, 2, 3, 4}
# Expected Output set()

s = {1, 2, 3, 4}

s.clear()

print(s)
```

```python
# 15. Write a Python program to create a frozen set and demonstrate its usage.
# Input [1, 2, 3, 4]
# Expected Output frozenset({1, 2, 3, 4})
# Modification is not allowed.

numbers = [1, 2, 3, 4]

s = frozenset(numbers)

print(s)
```

```python
# 16. Write a Python program to find common unique words from two sentences using sets.
# Input Sentence 1: python is easy and powerful
# Sentence 2: python is popular and useful
# Expected Output Common words = {'python', 'is', 'and'} (order may vary)

sentence1 = "python is easy and powerful"
sentence2 = "python is popular and useful"

set1 = set(sentence1.split())
set2 = set(sentence2.split())

common_words = set1.intersection(set2)

print("Common words =", common_words)
```
