```python
# 16. Write a Python program to find common unique words from two sentences using sets.
# Input Sentence 1: python is easy and powerful
# Sentence 2: python is popular and useful
# Expected Output Common words = {'python', 'is', 'and'} (order may vary)

sentence1 = "python is easy and powerful"
sentence2 = "python is popular and useful"

set1 = set(sentence1.split())
set2 = set(sentence2.split())

common_words = set1.intersection(set2)

print("Common words =", common_words)
```

```python
# 1. Write a Python program to find unique vowels present in a sentence using a set.
# Input Education is a powerful tool
# Expected Output Vowels = {'a', 'e', 'i', 'o', 'u'} (order may vary)

sentence = "Education is a powerful tool"

vowels = set()

for char in sentence.lower():
    if char in "aeiou":
        vowels.add(char)

print("Vowels =", vowels)
```

```python
# 4. Write a Python program to find missing numbers from 1 to n using set operations.
# Input Set = {1, 2, 4, 5, 7, 10}
# n = 10
# Expected Output Missing numbers = {3, 6, 8, 9} (order may vary)

s = {1, 2, 4, 5, 7, 10}
n = 10

numbers = set(range(1, n + 1))

missing = numbers - s

print("Missing numbers =", missing)
```

```python
# 5. Write a Python program to calculate the Jaccard similarity of two sets.
# Input A = {1, 2, 3, 4}
# B = {3, 4, 5, 6}
# Expected Output Jaccard similarity = 2 / 6 = 0.3333

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

intersection = A.intersection(B)
union = A.union(B)

jaccard = len(intersection) / len(union)

print("Jaccard similarity =", round(jaccard, 4))
```

```python
# 6. Write a Python program to find duplicate words in a sentence using sets.
# Input python is easy and python is powerful
# Expected Output Duplicate words = {'python', 'is'} (order may vary)

sentence = "python is easy and python is powerful"

words = sentence.split()

seen = set()
duplicate = set()

for word in words:
    if word in seen:
        duplicate.add(word)
    else:
        seen.add(word)

print("Duplicate words =", duplicate)
```

```python
# 7. Write a Python program to find all unique non-space characters in a string and display them in sorted order.
# Input data science
# Expected Output Unique characters = ['a', 'c', 'd', 'e', 'i', 'n', 's', 't']

text = "data science"

unique = set()

for char in text:
    if char != " ":
        unique.add(char)

result = sorted(unique)

print("Unique characters =", result)
```

```python
# 9. Write a Python program to create a set containing squares of all even numbers from 1 to 10.
# Input No user input required.
# Expected Output {4, 16, 36, 64, 100} (order may vary)

squares = set()

for i in range(1, 11):
    if i % 2 == 0:
        squares.add(i ** 2)

print(squares)
```

```python
# 10. Write a Python program to find characters that are common to every word in a list.
# Input Words = ['bella', 'label', 'roller']
# Expected Output Common characters = {'e', 'l'}

words = ['bella', 'label', 'roller']

common = set(words[0])

for word in words[1:]:
    common = common.intersection(set(word))

print("Common characters =", common)
```
