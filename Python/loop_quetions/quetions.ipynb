```python
"""Write a Python program to check whether a number is palindrome or not."""

n = int(input("Enter the number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Frequency of each digit in an integer
n = int(input("Enter the number: "))

for i in range(10):
    count = 0
    temp = n

    while temp > 0:
        digit = temp % 10

        if digit == i:
            count += 1

        temp //= 10

    if count > 0:
        print(i, "occurs", count, "times")
```

```python
#21. Find power of a number using for loop
base = int(input("Enter the base: "))
power = int(input("Enter the power: "))

result = 1

for i in range(power):
    result *= base

print("Answer =", result)
```

```python
#23. Find HCF (GCD) of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

hcf = 1

for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        hcf = i

print("HCF =", hcf)
```

```python
#Find LCM of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

lcm = max(a, b)

while True:
    if lcm % a == 0 and lcm % b == 0:
        break
    lcm += 1

print("LCM =", lcm)
```

```python
n = int(input("Enter the number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```
