```python
"""Write a Python program to check whether a number is palindrome or not."""

n = int(input("Enter the number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
