```python
"""Write a Python program to check whether a number is palindrome or not."""

n = int(input("Enter the number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Frequency of each digit in an integer
n = int(input("Enter the number: "))

for i in range(10):
    count = 0
    temp = n

    while temp > 0:
        digit = temp % 10

        if digit == i:
            count += 1

        temp //= 10

    if count > 0:
        print(i, "occurs", count, "times")
```

```python
#23. Find HCF (GCD) of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

hcf = 1

for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        hcf = i

print("HCF =", hcf)
```

```python
#Find LCM of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

lcm = max(a, b)

while True:
    if lcm % a == 0 and lcm % b == 0:
        break
    lcm += 1

print("LCM =", lcm)
```

```python
#Write a Python program to check whether a number is Armstrong number
#or not.
n = int(input("Enter the number: "))

original = n
digits = 0
temp = n

# Count number of digits
while temp > 0:
    digits += 1
    temp //= 10

sum = 0
temp = n

# Calculate sum of powers of digits
while temp > 0:
    digit = temp % 10
    sum += digit ** digits
    temp //= 10

if sum == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
n = int(input("Enter the value of n: "))

for num in range(1, n + 1):

    original = num
    temp = num
    digits = 0

    # Count digits
    while temp > 0:
        digits += 1
        temp //= 10

    # Calculate Armstrong sum
    temp = num
    sum = 0

    while temp > 0:
        digit = temp % 10
        sum += digit ** digits
        temp //= 10

    # Check Armstrong
    if sum == original:
        print(num, end=" ")
```

```python
#Write a Python program to print Fibonacci series up to n terms.
n = int(input("Enter the number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")

    c = a + b
    a = b
    b = c
```
